## Yengil kod orqali tushuntirish


### Kirishlar, Og'irliklar va Biriktirma (Input, Weights, and Bias):

```python
x = [8, 3]
w = [0.4, -0.6]
b = -1

z = w[0]*x[0] + w[1]*x[1] + b
print(z)
```

 x = [8, 3]: Bu kirish vektorini ifodalaydi. Neyronning qabul qiladigan ma'lumotlari.

 w = [0.4, -0.6]: Bu og'irliklar vektoridir. Har bir kirishning qanchalik muhimligini ko'rsatadi.

 b = -1: Bu biriktirma (bias) qiymati. Neyronning faollashuv chegarasini sozlashga yordam beradi.

 Bu yerda kirishlar (x) ularning og'irliklariga (w) ko'paytirilib, natijalar yig'indisi topiladi va bunga biriktirma (b) qo'shiladi. Bu neyronning kirish ma'lumotlaridan olingan umumiy yig'indisini ifodalaydi. Bizning holatimizda z ning qiymati 0.40000000000000036 chiqqan.

### **Faollashtirish Funksiyasi (Activation Function - Step Function):**

```python
natija = 1 if z >= 0 else 0
print(natija)
```

 natija = 1 if z >= 0 else 0: Bu sodda faollashtirish funksiyasi bo'lib, agar z ning qiymati noldan katta yoki unga teng bo'lsa, natija 1 ga, aks holda 0 ga teng bo'ladi. Bu neyronning oddiygina ikkilik qaror qabul qilishini ko'rsatadi (masalan, 'ha' yoki 'yo'q', 'faol' yoki 'faol emas'). Bizning holatimizda z musbat bo'lgani uchun natija 1 bo'lgan.

### **Sigmoid Faollashtirish Funksiyasi (Sigmoid Activation Function):**

```python
import numpy as np

def sigmoid(z):
  return 1/(1+np.exp(-z))

print(sigmoid(z))
```

 Bu sigmoid funksiyasi bo'lib, u z qiymatini 0 va 1 oralig'iga siqib chiqaradi. Natija 0.5 dan katta bo'lsa, u '1' ga yaqin, 0.5 dan kichik bo'lsa, '0' ga yaqin deb talqin qilinishi mumkin. Bu ko'pincha ehtimollikni hisoblashda yoki neyronning yumshoqroq (diskret bo'lmagan) chiqishini olishda ishlatiladi. Bizning holatimizda sigmoid(z) ning qiymati 0.5986876601124521 chiqqan, bu z ning musbat ekanligini tasdiqlaydi va neyronning 'faol' bo'lish ehtimoli yuqori ekanligini ko'rsatadi.

Deep learning (chuqur o'rganish) – bu mashinani o'rganishning bir sohasi bo'lib, inson miyasining neyron tarmoqlari tuzilishidan ilhomlangan sun'iy neyron tarmoqlaridan foydalanadi. U ma'lumotlardagi murakkab naqshlarni o'rganishga qodir bo'lib, katta hajmdagi ma'lumotlar bilan ishlashda juda samaralidir.

Deep learningning asosiy qismlari va tushunchalari quyidagilardan iborat:

Sun'iy Neyron Tarmoqlari (Artificial Neural Networks - ANNs):

Neyronlar (Neurons): Bular tarmoqning asosiy 'qurilish bloklari' bo'lib, kirish ma'lumotlarini oladi, ularni qayta ishlaydi va chiqishni hosil qiladi.

Qatlamlar (Layers): Neyronlar qatlamlarga bo'linadi:
Kirish qatlami (Input Layer): Tarmoqqa dastlabki ma'lumotlar kiritiladi.

Yashirin qatlamlar (Hidden Layers): Kirish va chiqish qatlamlari orasida joylashgan qatlamlar. Deep learning tarmoqlarida bu qatlamlar soni va murakkabligi ko'p bo'ladi, shuning uchun 'chuqur' deb ataladi.
Chiqish qatlami (Output Layer): Tarmoqning yakuniy natijasi (bashorat, tasnif, va hokazo) shu qatlamdan chiqadi.
Og'irliklar (Weights) va Biriktirmalar (Biases):

Og'irliklar (Weights - w): Neyronlar orasidagi bog'lanishlarning kuchini belgilaydi. O'qitish jarayonida tarmoq og'irliklarni moslashtiradi.

Biriktirmalar (Biases - b): Har bir neyronning chiqishiga qo'shiladigan qo'shimcha qiymat bo'lib, neyronning faollashuv chegarasini sozlashga yordam beradi.

Faollashtirish Funksiyalari (Activation Functions):

Neyronning kirish ma'lumotlarini qabul qilib, uni keyingi qatlamga uzatilishidan oldin ma'lum bir diapazonga siqib chiqaradi yoki nolga aylantiradi. Misollar: ReLU, Sigmoid, Tanh.
O'qitish (Training):

Oldinga Tarqatish (Forward Propagation): Kirish ma'lumotlari tarmoq orqali o'tkaziladi va chiqish hosil qilinadi.

Yo'qotish Funksiyasi (Loss Function): Tarmoqning bashorati bilan haqiqiy qiymat o'rtasidagi farqni o'lchaydi. Maqsad bu farqni minimallashtirish.

Orqaga Tarqatish (Backpropagation): Yo'qotish funksiyasidan kelgan xato signali tarmoqning orqasiga tarqatiladi va og'irliklar hamda biriktirmalar mos ravishda yangilanadi.

Optimallash Algoritmlari (Optimizers): Og'irliklar va biriktirmalarni yangilash jarayonini boshqaradi. Misollar: Gradient Descent, Adam, RMSprop.

Turlari (Types of Deep Learning Networks):

Konvolyutsion Neyron Tarmoqlar (Convolutional Neural Networks - CNNs): Tasvirni aniqlash va ishlov berishda juda samarali.
Takroriy Neyron Tarmoqlar (Recurrent Neural Networks - RNNs): Ketma-ket ma'lumotlar (masalan, matn, nutq) bilan ishlash uchun ishlatiladi.

Transformatorlar (Transformers): Asosan tabiiy tilni qayta ishlashda (NLP) eng so'nggi yutuqlarga erishgan tarmoqlar (masalan, ChatGPT).
Generativ Adversar Tarmoqlar (Generative Adversarial Networks - GANs): Yangi ma'lumotlar (rasmlar, videolar) yaratish uchun ishlatiladi.

Deep learning ko'plab sohalarda, jumladan, tasvirni aniqlash, nutqni tanish, tabiiy tilni qayta ishlash, tavsiya tizimlari va avtonom transport vositalarida inqilobiy o'zgarishlarga olib keldi.



GPU (Graphics Processing Unit): Dastlab grafika va tasvirlarni tez ishlash uchun mo'ljallangan edi, ammo hozirda parallel hisoblashdagi yuqori samaradorligi tufayli Deep Learning (Chuqur O'rganish) va boshqa ilmiy hisob-kitoblarda keng qo'llaniladi. U bir vaqtning o'zida ko'plab kichik operatsiyalarni bajarishga qodir.

TPU (Tensor Processing Unit): Google tomonidan maxsus ravishda neyron tarmoqlaridagi operatsiyalarni (ayniqsa, tensor operatsiyalarini) tezlashtirish uchun ishlab chiqilgan maxsus protsessor. Ular Machine Learning (Mashinani O'rganish) ish yuklarini, ayniqsa TensorFlow freymvorkida, GPU'larga qaraganda ham tezroq va energiya tejamkor ishlashga optimallashtirilgan.
