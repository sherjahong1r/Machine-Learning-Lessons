# **Feature extraction**

```python
import librosa
import pandas as pd
import os
from scipy.stats import kurtosis, mode
import IPython.display as ipd
```

```python
def extract_mean_frequency(y, sr=22050):
  return librosa.feature.spectral_centroid(y=y, sr=sr).mean()

# y: Bu audio vaqt seriyasi (audio signalning o'zi).
# sr: Bu audioning namuna olish tezligi (sekundiga nechta namuna, sukut bo'yicha 22050).
# librosa.feature.spectral_centroid(y=y, sr=sr): Bu qism audio signalning har bir freymi uchun spektral
# markazni hisoblaydi. Spektral markaz tovush spektrining 'massalar markazi' qayerda joylashganligin
# i ko'rsatuvchi o'lchovdir. U ko'pincha tovushning 'yorqinligi' yoki 'tembrini' tavsiflash uchun ishlatiladi.
# .mean(): Turli freymlar uchun spektral markaz hisoblangandan so'ng, bu barcha qiymatlarning
#  o'rtachasini oladi va butun audio klip uchun bitta o'rtacha spektral markaz qiymatini beradi.
```

```python
audio_file = "/content/audio (1).mp3"
```

## Standart Devision

```python
def axtract_standart_deviation(y, sr):
  return librosa.feature.spectral_centroid(y=y, sr=sr).std()
```

```python
devision = axtract_standart_deviation(y, sr)
devision
```

## Median

```python
def extract_median(y, sr):
  import numpy as np
  spectral_centroid_values = librosa.feature.spectral_centroid(y=y, sr=sr)[0]
  return np.median(spectral_centroid_values)
```

```python
# Get the spectral centroid values, which are typically returned as a 2D array, so we access the first row.
spectral_centroid_values = librosa.feature.spectral_centroid(y=y, sr=sr)[0]

# Calculate the median using numpy's median function.
median = np.median(spectral_centroid_values)
median
```

## Kurtosis

```python
def extract_kurtosis(y):
  return kurtosis(y)
```

## Va qolgan extract featurelar

```python
def extract_spectral_flatness(y, sr):
  return librosa.feature.spectral_flatness(y=y)[0]

def extract_mode(y):
  return mode(y)[0]

def extract_centrold(y, sr):
  return librosa.feature.spectral_centroid(y=y, sr=sr)[0]

def extract_peak(y, sr):
  return librosa.feature.spectral_rolloff(y=y, sr=sr)[0]

def extract_mean_function(y, sr):
  return librosa.feature.mfcc(y=y, sr=sr).mean()

def extract_maxsimum_function(y, sr):
  return librosa.feature.mfcc(y=y, sr=sr).max()

def extract_minimum_function(y, sr):
  return librosa.feature.mfcc(y=y, sr=sr).min()

def extract_mean_dom(y, sr):
  return librosa.feature.spectral_bandwidth(y=y, sr=sr).mean()

def extract_minimum_dom(y, sr):
  return librosa.feature.spectral_bandwidth(y=y, sr=sr).min()

def extract_maxsimum_dom(y, sr):
  return librosa.feature.spectral_bandwidth(y=y, sr=sr).max()

def extract_dfrange(maxdom, mindom):
  return maxdom - mindom

def extract_modindx(y, sr):
  return librosa.feature.tempogram(y=y, sr=sr).mean()
```

```python
def extract_features(audio_file):
    y, sr = librosa.load(audio_file)

    meanfreq = extract_mean_frequency(y, sr)
    sd = axtract_standart_deviation(y, sr)
    median = extract_median(y, sr)
    kurt = extract_kurtosis(y)
    sfm = extract_spectral_flatness(y, sr)
    mode_ = extract_mode(y)
    centroid = extract_centrold(y, sr)
    peakf = extract_peak(y, sr)
    meanfun = extract_mean_function(y, sr)
    minfun = extract_minimum_function(y, sr)
    maxfun = extract_maxsimum_function(y, sr)
    meandom = extract_mean_dom(y, sr)
    mindom = extract_minimum_dom(y, sr)
    maxdom = extract_maxsimum_dom(y, sr)
    dfrange = extract_dfrange(maxdom, mindom)
    modindx = extract_modindx(y, sr)

    features = {
        'meanfreq': meanfreq,
        'sd': sd,
        'median': median,
        'kurt': kurt,
        'sfm': sfm,
        'mode': mode_,
        'centroid': centroid,
        'peakf': peakf,
        'meanfun': meanfun,
        'minfun': minfun,
        'maxfun': maxfun,
        'meandom': meandom,
        'mindom': mindom,
        'maxdom': maxdom,
        'dfrange': dfrange,
        'modindx': modindx
    }

    return features

# Bu extract_features funksiyasi bitta audio faylni (audio_file) qabul qiladi. U avval librosa yordamida audioni yuklaydi
# (y, sr), so'ngra bir qancha oldindan belgilangan funksiyalar (masalan, extract_mean_frequency, axtract_standart_deviation
# va boshqalar) yordamida audio signalidan turli xil akustik xususiyatlarni (o'rtacha chastota, standart og'ish, mediana,
# kurtosis, spektral tekislik va hokazolar) ajratib oladi. Nihoyat, barcha ajratilgan xususiyatlarni lug'at (features) ko'rinishida qaytaradi.
```

```python
def process_audio_directory(input_dir, output_file):
    # Initialize an empty list to store features
    features_list = []

    # Loop through all audio files in the directory
    for filename in os.listdir(input_dir):
        if filename.endswith('.mp3'):
            audio_file = os.path.join(input_dir, filename)
            features = extract_features(audio_file)
            features_list.append(features)

    # Convert the list of features to a DataFrame
    df = pd.DataFrame(features_list)

    # Save the DataFrame to a CSV file
    df.to_csv(output_file, index=False)


# Example usage:
input_directory = 'clips/'
output_csv_file = 'output_features_new.csv'
process_audio_directory(input_directory, output_csv_file)

# Bu process_audio_directory funksiyasi ko'rsatilgan papkadagi (input_dir) barcha .mp3 fayllarni qidirib topadi.
# Har bir audio fayldan extract_features funksiyasi yordamida akustik xususiyatlarni ajratib oladi. So'ngra,
# bu xususiyatlarni yig'ib, Pandas DataFramega aylantiradi va uni belgilangan CSV fayliga (output_file) saqlaydi.
```
