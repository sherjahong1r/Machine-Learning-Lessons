Chuqur o'rganish (Deep Learning) - bu sun'iy intellektning, xususan mashinani o'rganishning (Machine Learning) bir sohasi bo'lib, inson miyasining tuzilishi va ishlash prinsipiga asoslangan ko'p qatlamli sun'iy neyron tarmoqlaridan foydalanadi.

Qisqacha ta'rifi:

Asosiy g'oya: Neyron tarmoqlarining juda ko'p qatlamlaridan foydalanish (shu sababli 'chuqur' deb nomlanadi) ma'lumotlardagi murakkab naqshlar va abstraksiyalarni avtomatik ravishda o'rganish.

Qanday ishlaydi: Har bir qatlam kirish ma'lumotlaridan yuqoriroq darajadagi xususiyatlarni (feature) ajratib oladi. Masalan, tasvirni qayta ishlashda birinchi qatlam chekkalarni, keyingi qatlam shakllarni, undan keyingisi esa ob'ektlarni tanib olishi mumkin.

O'rganish usuli: Tarmoq o'zining 'og'irliklari' (weights) va 'bias'larini (qiyshayishlarini) katta hajmdagi ma'lumotlar yordamida o'zgartirib, aniqroq natijalar berishga intiladi. Bu odatda orqaga tarqalish (backpropagation) algoritmi yordamida amalga oshiriladi.

Afzalliklari: Ma'lumotlarning avtomatik xususiyatlarini ajratish (feature extraction) qobiliyati, katta hajmdagi ma'lumotlar bilan ishlashda yuqori samaradorlik, murakkab masalalarni hal qilish (masalan, tasvirni aniqlash, nutqni tanib olish, tabiiy tilni qayta ishlash).

Misollar: Yuzni tanib olish, ovozli yordamchilar (Siri, Google Assistant), avtonom avtomobillar, tibbiy diagnostika.


`/////////////`

Neyron (yoki sun'iy neyron) - bu neyron tarmoqlarning asosiy hisoblash birligi hisoblanadi. U inson miyasidagi biologik neyronlarga o'xshatib yaratilgan.

Qisqacha ta'rifi:

Kirishlar (Inputs): Boshqa neyronlardan yoki tashqi muhitdan ma'lumot qabul qiladi. Har bir kirish o'zining 'og'irligi' (weight) bilan bog'liq bo'ladi.

Og'irliklar (Weights): Har bir kirishga berilgan ahamiyatni belgilaydigan sonlar. Bu og'irliklar neyronning o'rganish jarayonida moslashtiriladi.

Summa funksiyasi (Summation Function): Barcha kirishlarni ularning tegishli og'irliklariga ko'paytiradi va barcha natijalarni jamlaydi.
Aktivatsiya funksiyasi (Activation Function): Yig'indini qabul qiladi va neyronning yakuniy chiqishini (output) hosil qiladi. Bu funksiya neyronning 'yoqilishi' yoki 'o'chirilishi'ga qaror qiladi. Mashhur aktivatsiya funksiyalariga sigmoidal, ReLU, tanh kiradi.
Chiqish (Output): Aktivatsiya funksiyasidan keyin hosil bo'ladigan yakuniy qiymat, u keyingi neyronlarga yoki tarmoqning yakuniy chiqishiga uzatiladi.

Neyronlarning bir-biriga bog'lanishi va ko'p qatlamlarda joylashishi orqali murakkab naqshlarni o'rganish va ma'lumotlarni qayta ishlash mumkin.



`////////////////`

Perceptron (Perseptron) - bu neyron tarmoqlarning eng sodda shakllaridan biri bo'lib, sun'iy neyron tarmog'ining asosiy qurilish bloki hisoblanadi. U dastlab 1957-yilda Frank Rosenblatt tomonidan kiritilgan.

Qisqacha ta'rifi:

Vazifasi: Ikki sinfga ajratish (binary classification). Ya'ni, kirish ma'lumotlarini qaysidir belgi asosida ikki guruhga bo'lish.
Ishlash prinsipi: Bir nechta kirish qiymatlarini (inputlar) qabul qiladi, ularni tegishli og'irliklar (weights) bilan ko'paytiradi va barcha natijalarni jamlaydi. So'ngra, bu yig'indini aktivatsiya funksiyasidan (odatda, qadam funksiyasi) o'tkazib, chiqish (output) hosil qiladi.

O'rganish: Perceptron xato asosida o'rganadi. Agar u noto'g'ri tasniflansa, og'irliklarni shunday o'zgartiradiki, keyingi safar to'g'ri natija berishi mumkin.

Cheklovlar: Perceptron faqat chiziqli ajratiladigan (linearly separable) ma'lumotlar bilan ishlashga qodir. Ya'ni, agar siz kirish ma'lumotlarini tekis chiziq (yoki ko'p o'lchovli fazoda gipertekislik) bilan ikki guruhga ajrata olsangiz, u samarali ishlaydi. Masalan, XOR funksiyasini hal qila olmaydi.
Zamonaviy chuqur o'rganish modellarida Perceptronlar kamdan-kam holda yakka holda ishlatiladi, ammo ular murakkab neyron tarmoqlarining asosiy g'oyasini tashkil qiladi.



`//////////////////`

Aktivatsiya funksiyasi (Activation Function) - bu neyron tarmoqdagi neyronning yig'indisini (kirishlar va og'irliklar ko'paytmasi yig'indisini) qabul qilib, uning chiqishini aniqlaydigan matematik funksiyadir. Oddiy qilib aytganda, u neyronning 'yoqilishi' (faollashishi) yoki 'o'chirilishi'ni hal qiladi. Aktivatsiya funksiyalari neyron tarmoqlarga chiziqli bo'lmagan (non-linear) munosabatlarni o'rganish imkonini beradi, bu esa ularning murakkab vazifalarni bajarishi uchun juda muhimdir.

Eng keng tarqalgan aktivatsiya funksiyalariga quyidagilar kiradi:

Sigmoid: Chiqish qiymatini 0 va 1 oralig'iga siqib chiqaradi. Ko'pincha ikkilik tasniflash (binary classification) vazifalarida ishlatiladi, ammo gradiyent yo'qolishi (vanishing gradient) muammosiga moyil.

ReLU (Rectified Linear Unit): Agar kirish musbat bo'lsa, xuddi shu qiymatni qaytaradi, aks holda 0 ni qaytaradi. Eng mashhur funksiyalardan biri bo'lib, hisoblash tezligi va gradiyent yo'qolishi muammosini kamaytirishga yordam beradi.

Leaky ReLU: ReLU ning o'zgartirilgan versiyasi bo'lib, kirish manfiy bo'lganda juda kichik musbat qiymatni qaytaradi, bu esa 'o'lik' ReLU muammosini hal qilishga yordam beradi.

Tanh (Hyperbolic Tangent): Chiqish qiymatini -1 va 1 oralig'iga siqib chiqaradi. Sigmoidga o'xshash, lekin chiqish o'rtacha nolga yaqinroq bo'lgani uchun ba'zi hollarda yaxshiroq ishlashi mumkin.

Softmax: Ko'p klassli tasniflash (multi-class classification) vazifalarida oxirgi qatlamda ishlatiladi. Chiqish qiymatlarini yig'indisi 1 ga teng bo'lgan ehtimollik taqsimotiga aylantiradi.


`/////////////////`

Yo'qotish funksiyasi (Loss Function):

Ta'rif: Bu modelning bashorati (prognozi) bilan haqiqiy qiymat (yoki maqsad qiymat) o'rtasidagi farqni yoki xatoning miqdorini o'lchaydigan funksiya.

Maqsad: Yo'qotish funksiyasining qiymati qancha kichik bo'lsa, modelning ishlashi shuncha yaxshi bo'ladi.

Ishlashi: O'qitish jarayonida modelning og'irliklarini (weights) sozlash uchun qanchalik katta xato borligini aniqlashda ishlatiladi.

Misollar: Regressiya vazifalari uchun Mean Squared Error (MSE), klassifikatsiya vazifalari uchun Cross-Entropy Loss (Kesish-entropiya yo'qotishi) kabi funksiyalar mavjud.

Optimallashtirish (Optimization):

Ta'rif: Bu yo'qotish funksiyasining qiymatini minimallashtirish uchun modelning ichki parametrlarini (masalan, og'irliklar va biaslar) tizimli ravishda o'zgartirish jarayoni.

Maqsad: Modelni ma'lumotlarga eng yaxshi mos keladigan holatga keltirish, ya'ni yo'qotish funksiyasini eng past darajaga tushirish.

Ishlashi: Optimallashtirish algoritmlari (masalan, Gradient Descent, Adam, RMSprop) yo'qotish funksiyasining gradiyentini (o'zgarish yo'nalishini) hisoblab chiqadi va shu yo'nalishda parametrlarni o'zgartirib, funksiya qiymatini bosqichma-bosqich kamaytiradi.

Ahamiyati: Optimallashtirish jarayoni tufayli neyron tarmoqlar o'rganadi va yanada aniq bashorat qila oladi.

## Jadvalimiz quyidagicha ko'rinishga ega bo'lsin

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'BMI':[22,30,25,33],
    'Qon shakari(mg/dl)':[85,110,95,120],
    'Diabet':["0(Sog'lom)","1(Xavf ostida)","0(Sog'lom)","1(Xavf ostida)"]

}
df = pd.DataFrame(data)
df

# df nomli jadval yaratildi
```

BMI (Body Mass Index): Tana vaznining bo'yga nisbatan qancha ekanligini ko'rsatuvchi indeks. Bu ko'rsatkich insonning vazni normal, ortiqcha yoki kamligini baholash uchun ishlatiladi va diabet xavfi bilan bog'liq bo'lishi mumkin.

Qon shakari (mg/dl): Qondagi glyukoza miqdorini miligramm/detsilitr o'lchov birligida ifodalaydi. Yuqori qon shakar darajasi diabetning asosiy belgilaridan biridir.

Diabet: Ushbu ustun diabet kasalligining mavjudligini yoki xavfini ko'rsatadi. Bizning ma'lumotlarimizda "0(Sog'lom)" sog'lom insonni, "1(Xavf ostida)" esa diabet xavfi bo'lgan insonni bildiradi.



## Maqsad:

### `Diabet` kasaligini hafini tahmin qiluvchi qiluvchi modelni yaratamiz

```python
# df.style.hide(axis="index")
# df.style.set_properties(**{'text-align': 'left'}).hide(axis="index")
df.reset_index(drop=True, inplace=True)
# indexni o'chirayabmiz
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
# Bular PyTorch kutubxonasining asosiy qismlari bo'lib, tensorlar bilan ishlash,
# neyron tarmoqlarni qurish va ularni optimallashtirish uchun ishlatiladi.
import matplotlib.pyplot as plt
# Grafik chizish uchun

X = torch.tensor([[0.2, 0.3],
                  [0.6, 0.8],
                  [0.4, 0.5],
                  [0.7, 0.9]],
                  dtype=torch.float32
                  )

# X = torch.tensor(...): Bu sizning kirish ma'lumotlaringiz (features) hisoblanadi.
# Har bir qator bitta misolni (bemorning BMI va Qon shakari qiymatlari kabi) ifodalaydi.
# dtype=torch.float32 uning suzuvchi nuqtali sonlardan iborat ekanligini bildiradi.
# yani bunda normalizatsiya jarayoni amalga oshirildi malumotlar kam bo'lgani uchun shunday qilindi
# Buning o'rniga minmaxscaler yoki Standartscaler lardan ham foydalansak bo'lardi

y = torch.tensor([[0], [1], [0], [1]], dtype=torch.float32)

# y = torch.tensor(...): Bu esa bizning maqsad ma'lumotlarimiz (labels) hisoblanadi.
# Har bir qator X dagi mos misol uchun kutilgan natijani (masalan, diabet borligi yoki yo'qligi)
# ifodalaydi. Bu ham float32 turidagi sonlardan iborat.
```

```python
model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1),
    nn.Sigmoid()
)
# nn.Linear(2, 4): Bu birinchi to'liq bog'langan (fully connected) qatlam bo'lib, 2 ta
# kirish xususiyatini (BMI va Qon shakari) 4 ta chiqishga (neyronlarga) aylantiradi.
# nn.ReLU(): Bu aktivatsiya funksiyasi bo'lib, birinchi qatlamdan chiqqan natijalarga chiziqli
# bo'lmaganlik (non-linearity) qo'shadi. Agar qiymat musbat bo'lsa o'zini qaytaradi, manfiy bo'lsa 0 ga aylantiradi.
# nn.Linear(4, 1): Ikkinchi to'liq bog'langan qatlam bo'lib, oldingi qatlamdan chiqqan 4 ta natijani
# 1 ta yakuniy chiqishga (diabet bor/yo'qligi ehtimoli) aylantiradi.
# nn.Sigmoid(): Bu aktivatsiya funksiyasi modelning yakuniy chiqishini 0 va 1 oralig'iga siqib chiqaradi. Bu funksiya
# ikkilik tasniflash (binary classification) masalalari uchun ideal, chunki chiqish ehtimollik sifatida talqin qilinishi mumkin.
```

```python
loss = nn.BCELoss()  # Bu yuqorida torch dan import qilingan nn
# Modelning bashorati (y_pred) va haqiqiy qiymat (y) orasidagi farqni o'lchash uchun yo'qotish
# funksiyasini (Binary Cross-Entropy Loss) belgilaydi. Bu ikkilik tasniflash vazifalari uchun juda mos keladi.
optimizer = optim.SGD(model.parameters(), lr=0.05) # O'rganish tezligi 0.01 dan 0.05 ga oshirildi
# Modelning og'irliklarini (weights) sozlash uchun optimallashtiruvchini (Stochastic Gradient Descent)
# belgilaydi. lr (learning rate) o'rganish tezligini, ya'ni og'irliklar qanchalik tez yangilanishini nazorat qiladi.

epochs = 500 # Epochlar soni 100 dan 500 ga oshirildi
# Modelning butun ma'lumotlar to'plami bo'ylab necha marta o'qitilishini (takrorlanishini) belgilaydi.
losses = []

for epoch in range(epochs):          #Asosiy o'qitish sikli. Har bir epoch da:
    y_pred = model(X)                #Model X kirish ma'lumotlari asosida bashorat qiladi.
    loss_value = loss(y_pred, y)     #Bashorat va haqiqiy qiymat orasidagi yo'qotish hisoblanadi.
    losses.append(loss_value.item())

    optimizer.zero_grad()   # Oldingi qadamdagi gradientlarni tozalaydi.
    loss_value.backward()   # Yo'qotish funksiyasining gradientlarini hisoblaydi (orqaga tarqalish).
    optimizer.step()        # Hisoblangan gradientlar yordamida modelning og'irliklarini yangilaydi.

    if epoch % 50 == 0: # Chop etish chastotasi o'zgartirildi 10 dan 50 ga
    # Har 50 epoch da yo'qotish qiymatini chop etib, trening jarayonini kuzatishga yordam beradi.
        print(f"Epoch: {epoch}, Loss: {loss_value.item()}")

    # Chop etilgan natijalardan ko'rinib turibdiki, Loss qiymati Epochlar davomida asta-sekin kamayib bormoqda.
    # Masalan, 0-epochda Loss: 0.7055 bo'lsa, 450-epochga kelib Loss: 0.3238 ga tushgan. Bu shuni anglatadiki,
    # model ma'lumotlardan o'rganmoqda va uning bashoratlari haqiqiy qiymatlarga yaqinlashib bormoqda,
    # ya'ni modelning aniqligi yaxshilanmoqda.
```

```python
plt.plot(losses)
plt.xlabel("Epoxa")
plt.ylabel("Yo'qotish")

# Grafik ko'rinishida chiqarildi
```

```python
yangi_bemor = torch.tensor([[0.85, 0.75]], dtype=torch.float32)
y_pred = model(yangi_bemor)
print(y_pred.item())

# Bunda yangi bemor o'rgatildi va va uning Diabetga chalinishlik shansi yani ehtimoli 76 % ekan
# sigmoid borligi uchun ham avtomatiskiy foiz ko'rinishida chiqarib berilmoqda
```

Umumiy qisqacha xulosa shuki, modelimiz kichik ma'lumotlar to'plamiga qaramay, o'rganish jarayonida sezilarli yutuqlarga erishdi. Yo'qotish funksiyasining (loss) qiymati 0.68 dan 0.39 gacha kamayishi, modelning ma'lumotlardagi naqshlarni o'rganib, bashorat qilish qobiliyatini oshirganini ko'rsatadi. Oxirgi bosqichda esa yangi bemor uchun diabetga chalinish ehtimoli 76% deb baholandi, bu modelning haqiqiy dunyo ma'lumotlari bo'yicha ham bashorat bera olishini namoyish etadi. Lekin, yuqorida aytib o'tganimizdek, yanada aniqroq va ishonchli natijalar uchun kattaroq va xilma-xil ma'lumotlar to'plami talab qilinadi.

```python
data = {
    'BMI':[22,30,25,33],
    'Qon shakari(mg/dl)':[85,110,95,120],
    'Diabet':["0(Sog'lom)","1(Xavf ostida)","0(Sog'lom)","1(Xavf ostida)"]

}
df = pd.DataFrame(data)
df
```

```python
from sklearn.preprocessing import MinMaxScaler

# Scaler yaratish
scaler = MinMaxScaler()

# 'BMI' va 'Qon shakari(mg/dl)' ustunlarini normalizatsiya qilish
df[['BMI', 'Qon shakari(mg/dl)']] = scaler.fit_transform(df[['BMI', 'Qon shakari(mg/dl)']])

# Normalizatsiya qilingan jadvalni ko'rish
display(df)
```

Endi df jadvalidagi 'BMI' va 'Qon shakari(mg/dl)' ustunlari 0 dan 1 gacha bo'lgan oraliqda normalizatsiya qilingan. Keyin bu qiymatlarni torch.tensor ga o'tkazib, modelni qayta o'qitish mumkin bo'ladi.

```python
# Normalizatsiya qilingan qiymatlarni X tensoriga o'tkazish
X_normalized = torch.tensor(df[['BMI', 'Qon shakari(mg/dl)']].values, dtype=torch.float32)

# 'Diabet' ustunini 0 va 1 ga o'tkazish (agar hali qilinmagan bo'lsa)
# Masalan, '0(Sog'lom)' ni 0 ga, '1(Xavf ostida)' ni 1 ga o'tkazish
df['Diabet_num'] = df['Diabet'].apply(lambda x: 1 if '1(Xavf ostida)' in x else 0)
y_normalized = torch.tensor(df['Diabet_num'].values, dtype=torch.float32).unsqueeze(1)

print("Normalizatsiya qilingan X (kirish ma'lumotlari):")
print(X_normalized)
print("Normalizatsiya qilingan y (chiqish ma'lumotlari):")
print(y_normalized)
```

```python
model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1),
    nn.Sigmoid()
)
```

```python
loss = nn.BCELoss()
optimizer = optim.SGD(model.parameters(), lr=0.05)

epochs = 500
losses = []

for epoch in range(epochs):
    y_pred = model(X)
    loss_value = loss(y_pred, y)
    losses.append(loss_value.item())

    optimizer.zero_grad()
    loss_value.backward()
    optimizer.step()

    if epoch % 50 == 0:
        print(f"Epoch: {epoch}, Loss: {loss_value.item()}")
```
